```python
import pandas as pd 
from gensim.models import Word2Vec, FastText
from gensim.utils import simple_preprocess

from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
```

```python
sentences = ["Cat sat on the mat.", "The dog barked at the cat.", "The cat chased the mouse.", "The mouse ran away from the cat."]
tokenized_sentences = [simple_preprocess(sentence) for sentence in sentences]
```

```python
Word2Vec_model = Word2Vec(sentences=tokenized_sentences, vector_size=100, window=5, min_count=1, workers=4)
FastText_model = FastText(sentences=tokenized_sentences, vector_size=100, window=5, min_count=1, workers=4)
```

```python
import plotly.graph_objects as go
from sklearn.decomposition import PCA

def plot_embeddings(model, title):
    words = list(model.wv.index_to_key)
    word_vectors = model.wv[words]

    pca = PCA(n_components=3)
    reduced_vectors = pca.fit_transform(word_vectors)

    fig = go.Figure()

    fig.add_trace(go.Scatter3d(
        x=reduced_vectors[:, 0],
        y=reduced_vectors[:, 1],
        z=reduced_vectors[:, 2],
        mode='markers+text',
        text=words,
        textposition='top center',
        marker=dict(size=5)
    ))

    fig.update_layout(
        title=title,
        scene=dict(
            xaxis_title="PCA 1",
            yaxis_title="PCA 2",
            zaxis_title="PCA 3"
        ),
        width=900,
        height=700
    )

    fig.show()
```

```python
plot_embeddings(Word2Vec_model, "Word2Vec Embeddings")
plot_embeddings(FastText_model, "FastText Embeddings")
```
